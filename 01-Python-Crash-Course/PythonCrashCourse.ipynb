### while loop

```python
# while loop will iterate until the condition is false
counter = 4
while counter > 2:
    print(counter)
    counter = counter-1
```

### for loop

```python
# for loop is used when we want to iterate or repeat multiple times
# range is a function and takes 3 parameters - start index(its inclusive), end index(it's exclusive) and step size(default value is 1)
for x in range(1, 6):
    print(x)
```

```python
for x in range(1, 6, 2):
    print(x)
```

```python
# Write a program to with start x at 11, decrement till 2
for x in range(11, 1, -1):
    print(x)
```
